```python
import random

list_1 = [random.randint(1, 20) for _ in range(20)]
list_2 = [number for number in list_1 if number % 2 == 0]

list_3 = []
for number in list_2:
    if number not in list_3:
        list_3.append(number)

list_3.sort()

list_4 = list_3[1:-1]

result = 0
for i in list_4:
    if i % 3 == 0:
        result += i

print(result)
```

```python
def divide(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        print("division by zero!")
    else:
        print("result is", result)
    finally:
        print("executing finally clause")
divide(2, 1)
divide(2, 0)
divide("2", "1")
```

```python
def countdown(num):
    print('Starting')
    while num > 0:
        yield num
        num -= 1
for x in countdown(3):
    print(x)
```
