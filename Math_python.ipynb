```python
def get_mean(data):
    return sum(data)/len(data)
data = [4,5,6,7,8,244,5]
d=get_mean(data)
d
```

```python
def get_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    
    if n%2==1:
        return sorted_data[n//2]
    else:
        mid1= sorted_data[n//2-1]
        mid2= sorted_data[n//2]
        return (mid1+mid2)/2
data = [1,2,3,4,5,6,7,8,9,8,7,6,5,4,3,2,1]
d1 = get_median(data)
d1
```

```python
def get_mode(data):
    frequency = {}

    for item in data:
        if item in frequency:
            frequency[item] += 1
        else:
            frequency[item] = 1

    max_frequency = max(frequency.values())

    if max_frequency == 1:
        return "No mode"

    modes = []

    for item, count in frequency.items():
        if count == max_frequency:
            modes.append(item)

    if len(modes) == 1:
        return modes[0]

    return modes


n = int(input("Enter the number of elements: "))

data = []

for i in range(n):
    value = int(input(f"Enter element {i + 1}: "))
    data.append(value)

print("Mode =", get_mode(data))
```

```python
data = [10, 20, 20, 30, 40, 50]
```

```python
import statistics

print("Using statistics module")
print("Mean   =", statistics.mean(data))
print("Median =", statistics.median(data))

try:
    print("Mode   =", statistics.mode(data))
except statistics.StatisticsError:
    print("Mode   = No unique mode")

print("All Modes =", statistics.multimode(data))
print()
```

```python
import numpy as np

print("Using NumPy")
print("Mean   =", np.mean(data))
print("Median =", np.median(data))

values, counts = np.unique(data, return_counts=True)
max_count = np.max(counts)
modes = values[counts == max_count]

if max_count == 1:
    print("Mode   = No mode")
elif len(modes) == 1:
    print("Mode   =", modes[0])
else:
    print("Mode   =", list(modes))
print()
```

```python
import pandas as pd

s = pd.Series(data)

print("Using Pandas")
print("Mean   =", s.mean())
print("Median =", s.median())

mode_values = s.mode().tolist()

if len(mode_values) == len(set(data)):
    print("Mode   = No mode")
elif len(mode_values) == 1:
    print("Mode   =", mode_values[0])
else:
    print("Mode   =", mode_values)
print()
```

```python
from scipy import stats

print("Using SciPy")
print("Mean   =", stats.tmean(data))
print("Median =", np.median(data))

result = stats.mode(data, keepdims=True)
mode_value = result.mode[0]
mode_count = result.count[0]

if mode_count == 1:
    print("Mode   = No mode")
else:
    print("Mode   =", mode_value)
print()
```
