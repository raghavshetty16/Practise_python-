1. Write a Python function to check whether a number is prime or not.

```python
def is_prime_iterative(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
print(is_prime_iterative(5))
```

```python
def is_prime_recursive(n, i=2):
    if n <= 1:
        return False
    if i > int(n**0.5):
        return True
    if n % i == 0:
        return False
    return is_prime_recursive(n, i + 1)
print(is_prime_recursive(5))
```

2.Write a Python program to sort a list of tuples using Lambda.

```python
# Input list of tuples
a = [(1, 'Python'), (2, 'with'), (3, 'GFG')]

# Define the specific order
order = ['with', 'GFG', 'Python']

# Create a mapping of elements to their positions
order_map = {value: index for index, value in enumerate(order)}

# Sort the list of tuples using the mapping
res = sorted(a, key=lambda x: order_map[x[1]])
print(res)
```

```python
# 2. Sort tuples
def sort_tuples(lst):
    return sorted(lst, key=lambda x: x[1])
```

4. Write a Python function that takes two lists and returns True if they have at least one common member.

```python
def has_common(list1, list2):
    return bool(set(list1) & set(list2))


# Example
print(has_common([1, 2, 3], [3, 4, 5]))  # True
print(has_common([1, 2], [4, 5]))        # False
```

```python
def has_common(list1, list2):
    for item in list1:
        if item in list2:
            return True
    return False
print(has_common([1, 2, 3], [3, 4, 5]))  
print(has_common([1, 2], [4, 5]))   
```

5. Write a Python function to calculate the factorial of a number.

```python
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result


# Example
print(factorial(5))  # 120
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


# Example
print(factorial(5))  # 120
```

```python
import math
print(math.factorial(5))
```

6. Write a Python program to reverse a string.

```python
def reverse_string(s):
    return s[::-1]


# Example
print(reverse_string("hello"))  
```

```python
def reverse_string(s):
    rev = ""
    for char in s:
        rev = char + rev
    return rev


print(reverse_string("hello"))  # olleh
```

```python
def reverse_string(s):
    if len(s) == 0:
        return s
    return s[-1] + reverse_string(s[:-1])


print(reverse_string("hello"))  # olleh
```

```python
def reverse_string(s):
    return ''.join(reversed(s))


print(reverse_string("hello"))  # olleh
```

7. Write a Python function to calculate the Fibonacci series.

```python
def fibonacci(n):
    series = []
    a, b = 0, 1
    
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    
    return series


# Example
print(fibonacci(7))  
# Output: [0, 1, 1, 2, 3, 5, 8]
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)


# Print series
for i in range(7):
    print(fibonacci(i), end=" ")
```

8. Write a Python function that takes a list and returns a new list with unique elements of the first list.

```python
def unique_list(lst):
    return list(set(lst))


# Example
print(unique_list([1, 2, 2, 3, 4, 4]))
# Output: [1, 2, 3, 4]
```

```python
def unique_list(lst):
    result = []
    for item in lst:
        if item not in result:
            result.append(item)
    return result


print(unique_list([1, 2, 2, 3, 4, 4]))
# Output: [1, 2, 3, 4]
```

```python
def unique_list(lst):
    return list(dict.fromkeys(lst))


print(unique_list([1, 2, 2, 3, 4, 4]))
```

9. Write a Python program to check if a given key already exists in a dictionary.

```python
def key_exists(d, key):
    return key in d


# Example
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(key_exists(my_dict, 'b'))  # True
print(key_exists(my_dict, 'x'))  # False
```

```python
def key_exists(d, key):
    return key in d.keys()
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(key_exists(my_dict, 'b'))  # True
print(key_exists(my_dict, 'x'))
```

```python
def key_exists(d, key):
    return d.get(key) is not None
my_dict = {'a': 1, 'b': 2, 'c': 3}
print(key_exists(my_dict, 'b'))  # True
print(key_exists(my_dict, 'x'))
```

10. Write a Python program to find the highest 3 values in a dictionary.

```python
def top_three_values(d):
    return sorted(d.values(), reverse=True)[:3]


# Example
my_dict = {'a': 10, 'b': 45, 'c': 23, 'd': 67, 'e': 12}
print(top_three_values(my_dict))  
# Output: [67, 45, 23]
```

```python
def top_three_items(d):
    return sorted(d.items(), key=lambda x: x[1], reverse=True)[:3]

my_dict = {'a': 10, 'b': 45, 'c': 23, 'd': 67, 'e': 12}
print(top_three_items(my_dict))
# Output: [('d', 67), ('b', 45), ('c', 23)]
```

11. Write a Python function to create and print a list where the values are squares of numbers between 1 and n (both included).

```python
def squares_list(n):
    return [i**2 for i in range(1, n+1)]


# Example
print(squares_list(5))
# Output: [1, 4, 9, 16, 25]
```

```python
def squares_list(n):
    result = []
    for i in range(1, n+1):
        result.append(i*i)
    return result


print(squares_list(5))
```

```python
def print_squares(n):
    for i in range(1, n+1):
        print(i*i, end=" ")


print_squares(5)
```

12. Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```python
def count_case(s):
    upper = 0
    lower = 0
    
    for char in s:
        if char.isupper():
            upper += 1
        elif char.islower():
            lower += 1
    
    return upper, lower


# Example
result = count_case("Hello World")
print("Uppercase:", result[0])
print("Lowercase:", result[1])
```

```python
def count_case(s):
    result = {"upper": 0, "lower": 0}
    
    for char in s:
        if char.isupper():
            result["upper"] += 1
        elif char.islower():
            result["lower"] += 1
    
    return result


print(count_case("Hello World"))
```
