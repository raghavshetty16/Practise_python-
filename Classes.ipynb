```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Marks: {self.marks}")

student1 = Student("Raghav", 21, 85)
student1.display()
```

```python
class BankAccount:
    def __init__ (self, account_holder_name, account_number, balance):
        self.account_holder_name = account_holder_name
        self.account_number = account_number
        self.balance = balance
        
    def withdraw(self, amount, balance):
        if amount > balance:
            print("Insuffiecient funds")
        else:
            self.balance -= amount
    
    def deposit(self, amount):
        self.balance += amount
    
    def get_balance(self):
        return self.balance
    
```

```python
account1 = BankAccount("raghav", "123456789", 1000)

#how to save this such that the information is saved even file vs code is closed, without db or cloud storage??
```

# Saving information in file 

```python
import pickle

with open("account1.pkl", "wb") as file: #the file word is just a variable name
    pickle.dump(account1, file)
```

```python
# how to load the information from the file
import pickle

with open("account1.pkl", "rb") as f:
    account1 = pickle.load(f)
    
# to run this function we need to first execute the BankAccount class and then execute the code to load the information 
# from the file, otherwise we will get an error because the class definition is not available when we try to load the object.
```
